```python
from langchain_groq import ChatGroq
```

```python
import os
```

```python
model = "openai/gpt-oss-20b"
api_key = os.getenv["GROQ_API_KEY"]
```

```python
llm = ChatGroq([
    model="openai/gpt-oss-20b",
    api_key=api_key])
```

```python
llm.invoke("hii")
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
input_temp=ChatPromptTemplate.from_messages([
    ("system","you are a exprience  poet.write meaningful poem.do not response query other than poem"),
    ("user","write a poem on {poem_topic}"),
])
```

```python
input=input_temp.invoke({"poem_topic":"rain fall"})
```

```python
response=llm.invoke(input)
```

```python
response.content
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser = StrOutputParser()
```

```python
parser.invoke(response)
```

```python
input=input_temp.invoke({"poem_topic":"rain fall"})
response=llm.invoke(input)
final_output=parser.invoke(response)
print(final_output)
```

```python
chain=input_temp | llm | parser
```

```python
chain.invoke({"poem_topic":"rain fall"})
```

```python
from langchain_core.output_parsers import JsonOutputParser
```

```python
json_parser = JsonOutputParser()
```

```python
json_parser.get_format_instructions()
```

```python
summerize_temp = ChatPromptTemplate.from_messages([
    ("system", "You are a expert in summerizing poem. Summerize poem in 5 points {JSON_FORMATIMG}"),
    ("user", "summerize the poem {poem}"),
]).partial(JSON_FORMATIMG=json_parser.get_format_instructions())
```

```python
summerize_temp
```

```python
final_response = chain.invoke({"poem_topic":"rain fall"})
```

```python
summerize_temp.invoke({"poem":"Life is a journey, full of ups and downs. It teaches us lessons, shapes our character, and helps us grow. Embrace the challenges, cherish the moments, and keep moving forward."})
```

```python
final_response_1 = chain.invoke(step_1_response)
```
